```python
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Grad_project/Sign Language Graduation Project/letters_hand_landmarks.csv"
df = pd.read_csv(file_path)

print("Shape:", df.shape)

# 1. عرض الأعمدة ذات القيم الفريدة القليلة (غالباً عمود الحرف أو الفئة)
print("\n--- Candidate Label / Metadata Columns (nunique <= 100) ---")
for idx, col in enumerate(df.columns):
    nunq = df[col].nunique(dropna=False)
    if nunq <= 100:
        print(f"idx={idx:2d} | col={col!r:10s} | dtype={df[col].dtype} | nunique={nunq} | sample={df[col].iloc[0]!r}")

# 2. تحديد عمود الحرف تلقائياً إذا كان فيه 32 قيمة فريدة
label_candidates = [c for c in df.columns if df[c].nunique() == 32]
print("\nColumns with 32 unique values:", label_candidates)

if label_candidates:
    label_col = label_candidates[0]
    print("\n--- Class Counts ---")
    print(df[label_col].value_counts())

    # 3. فحص الأعمدة الرقمية والصفوف الصفرية
    num_cols = [c for c in df.columns if c != label_col and pd.api.types.is_numeric_dtype(df[c])]
    print("\nNumeric feature columns count:", len(num_cols))
    coords = df[num_cols].to_numpy(dtype=np.float32)
    all_zero_rows = np.all(np.isclose(coords, 0.0, atol=1e-7), axis=1).sum()
    print("All-zero rows:", all_zero_rows)
```

```python
import numpy as np
import pandas as pd

file_path = "/content/drive/MyDrive/Grad_project/Sign Language Graduation Project/letters_hand_landmarks.csv"
df = pd.read_csv(file_path)

print("Shape:", df.shape)
print("All columns:", list(df.columns))

# فحص الصفوف التي تكون فيها الأعمدة الرقمية (باستثناء 4190) أصفاراً
num_cols = [c for c in df.columns if c != "4190" and pd.api.types.is_numeric_dtype(df[c])]
coords = df[num_cols].to_numpy(dtype=np.float32)
zero_rows = np.all(np.isclose(coords, 0.0, atol=1e-7), axis=1).sum()
print("Numeric cols count:", len(num_cols))
print("All-zero rows:", zero_rows)
```
